```python
import os

train_dir = '/content/drive/My Drive/kisan_mitra_data/train'

# Map duplicate or numbered folders to their clean primary category
folder_mapping = {
    'blast (1)': 'blast',
    'healthy (1)': 'healthy',
    'insect (1)': 'insect',
    'leaf_folder (1)': 'leaf_folder',
    'scald (1)': 'scald',
    'stripes (1)': 'stripes',
    'tungro (1)': 'tungro'
}

print("Initialise class folder consolidation")

for old_name, new_name in folder_mapping.items():
    old_path = os.path.join(train_dir, old_name)
    new_path = os.path.join(train_dir, new_name)

    if os.path.exists(old_path):
        if not os.path.exists(new_path):
            os.makedirs(new_path)
        # Move all images from the duplicate folder into the clean primary folder
        for filename in os.listdir(old_path):
            src_file = os.path.join(old_path, filename)
            dst_file = os.path.join(new_path, filename)
            if os.path.isfile(src_file):
                os.rename(src_file, dst_file)
        # Remove the now-empty old folder
        os.rmdir(old_path)
        print(f"Merged {old_name} into {new_name}")

print("", os.listdir(train_dir))
```

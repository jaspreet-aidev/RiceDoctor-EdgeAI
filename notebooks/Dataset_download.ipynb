```python
from google.colab import drive
drive.mount('/content/drive')
print("System Status: Mounting Google Drive...")
```

```python
import os
import zipfile

# Define your target folder path in Google Drive
drive_folder = '/content/drive/My Drive/kisan_mitra_data'
os.makedirs(drive_folder, exist_ok=True)

# Locate your downloaded zip file (from Kaggle)
zip_filename = 'rice-leaf-disease-riceguard-19k-cleaned.zip'

if os.path.exists(zip_filename):
    print(f"System Status: Found {zip_filename}. Unzipping directly into {drive_folder}...")

    # Unzip the dataset straight into your target folder
    with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
        zip_ref.extractall(drive_folder)

    print("System Status: Unzipping complete! Your dataset is permanently secured inside your 'kisan_mitra_data' Google Drive folder.")
else:
    print(f"Error: Could not find '{zip_filename}'. Make sure you ran the Kaggle download command first!")
```
